# CSL448 – Computational Linguistics and NLP
## Experiment 7 – Chunking and Chinking

**Aim:** Apply chunking and chinking rules to extract noun phrases and verb phrases from raw sentences.

```python
# Install and import required libraries
!pip -q install nltk spacy

import nltk
import spacy

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

print("NLTK and spaCy imported successfully.")
```

## 1. Raw sentences and POS tagging

```python
sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "The students are studying natural language processing.",
    "The intelligent system analyzes the text carefully.",
    "Our professor explained the difficult concept clearly."
]

for sentence in sentences:
    tokens = nltk.word_tokenize(sentence)
    tagged = nltk.pos_tag(tokens)
    print("Sentence:", sentence)
    print("POS tags:", tagged)
    print()
```

## 2. Chunking grammar

Chunking groups POS-tagged words into shallow phrases.

- **NP:** determiner + adjectives + one or more nouns
- **VP:** verbs with optional adverbs or complements

```python
grammar = r'''
    NP: {<DT>?<JJ.*>*<NN.*>+}
    VP: {<MD><VB.*>}
        {<VB.*><RB.*>*}
'''

chunk_parser = nltk.RegexpParser(grammar)

for sentence in sentences:
    tagged = nltk.pos_tag(nltk.word_tokenize(sentence))
    tree = chunk_parser.parse(tagged)

    print("\nSentence:", sentence)
    print(tree)
```

## 3. Extract noun phrases (NP) and verb phrases (VP)

```python
def extract_phrases(tree):
    noun_phrases = []
    verb_phrases = []

    for subtree in tree.subtrees():
        if subtree.label() == "NP":
            noun_phrases.append(
                " ".join(word for word, tag in subtree.leaves())
            )
        elif subtree.label() == "VP":
            verb_phrases.append(
                " ".join(word for word, tag in subtree.leaves())
            )

    return noun_phrases, verb_phrases

for sentence in sentences:
    tagged = nltk.pos_tag(nltk.word_tokenize(sentence))
    tree = chunk_parser.parse(tagged)

    np_list, vp_list = extract_phrases(tree)

    print("Sentence:", sentence)
    print("Noun Phrases:", np_list)
    print("Verb Phrases:", vp_list)
    print()
```

## 4. Chinking

Chinking removes unwanted POS patterns from a chunk.

Here, prepositions (`IN`) and conjunctions (`CC`) are excluded from noun phrase chunks.

```python
chink_grammar = r'''
    NP: {<DT>?<JJ.*>*<NN.*>+}
        }<IN|CC>{
'''

chink_parser = nltk.RegexpParser(chink_grammar)

for sentence in sentences:
    tagged = nltk.pos_tag(nltk.word_tokenize(sentence))
    tree = chink_parser.parse(tagged)

    np_list, _ = extract_phrases(tree)

    print("\nSentence:", sentence)
    print("Noun Phrases after chinking:", np_list)
    print(tree)
```

## 5. Try your own sentence

```python
custom_sentence = "The experienced NLP student is building a language model."

tagged = nltk.pos_tag(nltk.word_tokenize(custom_sentence))
tree = chunk_parser.parse(tagged)

np_list, vp_list = extract_phrases(tree)

print("Sentence:", custom_sentence)
print("POS tags:", tagged)
print("Noun Phrases:", np_list)
print("Verb Phrases:", vp_list)
print("\nParse tree:")
print(tree)
```

## Conclusion

Chunking and chinking rules were successfully applied to POS-tagged raw sentences. Noun phrases and verb phrases were extracted using NLTK's `RegexpParser`.

- **Chunking** adds words to a phrase based on POS patterns.
- **Chinking** removes unwanted words/patterns from a chunk.
