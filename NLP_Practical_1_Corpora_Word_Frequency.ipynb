## 2. Explore the Brown Corpus

```python
from nltk.corpus import brown

words = brown.words()
print("Number of words:", len(words))
print("Number of sentences:", len(brown.sents()))
print("Number of categories:", len(brown.categories()))
print("Categories:", brown.categories())
print("\nFirst 30 words:")
print(words[:30])
```

## 3. Word-frequency distribution

```python
word_freq = Counter(word.lower() for word in words if word.isalpha())

print("Top 30 most frequent words:")
for word, count in word_freq.most_common(30):
    print(f"{word}: {count}")
```

## 4. Plot the frequency distribution

```python
top_words = word_freq.most_common(20)
labels = [word for word, count in top_words]
counts = [count for word, count in top_words]

plt.figure(figsize=(12, 5))
plt.bar(labels, counts)
plt.xticks(rotation=60)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 20 Words in the Brown Corpus")
plt.show()
```
