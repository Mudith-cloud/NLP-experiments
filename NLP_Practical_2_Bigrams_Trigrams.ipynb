## 1. Install and import libraries

```python
!pip -q install nltk spacy
import nltk
import spacy
from nltk.corpus import brown
from nltk.util import bigrams, trigrams
from collections import Counter

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
```

## 2. Load and prepare the corpus

```python
tokens = [word.lower() for word in brown.words() if word.isalpha()]

print("Number of cleaned tokens:", len(tokens))
print("First 30 tokens:")
print(tokens[:30])
```

## 3. Generate bigrams

```python
bigram_freq = Counter(bigrams(tokens))

print("Top 30 bigrams:")
for bg, count in bigram_freq.most_common(30):
    print(f"{bg}: {count}")
```

## 4. Generate trigrams

```python
trigram_freq = Counter(trigrams(tokens))

print("Top 30 trigrams:")
for tg, count in trigram_freq.most_common(30):
    print(f"{tg}: {count}")
```

## 5. Sort and display by frequency

```python
sorted_bigrams = sorted(bigram_freq.items(), key=lambda x: x[1], reverse=True)
sorted_trigrams = sorted(trigram_freq.items(), key=lambda x: x[1], reverse=True)

print("Top 10 sorted bigrams:")
for item, count in sorted_bigrams[:10]:
    print(item, "->", count)

print("\nTop 10 sorted trigrams:")
for item, count in sorted_trigrams[:10]:
    print(item, "->", count)
```
