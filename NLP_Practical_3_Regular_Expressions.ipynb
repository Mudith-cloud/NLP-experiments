# NLP Practical 3 — Regular Expressions for Information Extraction

**Google Colab | NLTK + spaCy**

## 1. Import regular expressions

```python
import re

text = '''
For project queries, contact rahul@gmail.com or priya.sharma@example.com.
You can also call +91-9876543210, 9876543210, or 9812345678.
Meeting ID: NLP-2026-08.
Website: https://example.com
Date: 07/08/2026
''' 

print(text)
```

## 2. Extract email IDs

```python
email_pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'

emails = re.findall(email_pattern, text)

print("Email IDs:")
for email in emails:
    print(email)
```

## 3. Extract Indian phone numbers

```python
phone_pattern = r'(?:\+91[-\s]?)?[6-9]\d{9}'

phones = re.findall(phone_pattern, text)

print("Phone numbers:")
for phone in phones:
    print(phone)
```

## 4. Extract dates

```python
date_pattern = r'\b(?:0?[1-9]|[12]\d|3[01])[/.-](?:0?[1-9]|1[0-2])[/.-](?:\d{2}|\d{4})\b'

dates = re.findall(date_pattern, text)

print("Dates:")
for date in dates:
    print(date)
```

## 5. Extract URLs

```python
url_pattern = r'https?://[^\s]+'

urls = re.findall(url_pattern, text)

print("URLs:")
for url in urls:
    print(url)
```

## 6. Extract custom IDs

```python
id_pattern = r'\b[A-Z]{2,}-\d{4}-\d{2}\b'

ids = re.findall(id_pattern, text)

print("Custom IDs:")
for item in ids:
    print(item)
```
